**Question 1:** How many customers are in the file?
(as help, we have added some comments and starter code to help you structure the solution)

```python
from os.path import exists
import csv

# check if file can be found and open the file
datafile = "./sample_data/customers.csv"
if not exists(datafile):
  raise SystemExit("You should run the first code cell and download the dataset files!")

# customers file is in the sample_data folder and can be opened.
# Using csv reader is one way of dealing with the file
# but it's up to you how you want to proceed
with open(datafile, 'r') as fl:
  csvreader = csv.reader(fl, delimiter=',')

  customers = [row for row in csvreader if row]

# 1. Total de clientes (filas)
print("1. Total customers:", len(customers))



print("All done!")
```

**Question 2:** In how many different states do the customers live in?

```python
# open file and init vars
states = set()

with open(datafile, "r") as fl:
  csvreader = csv.reader(fl, delimiter=",")
  next(csvreader)

  for row in csvreader:
    states.add(row[4])
# evaluate the contents of the file
print(f"Dif states:  {len(states)}")
```

**Question 3** What is the state with most customers?

```python
# open file and init vars
states_count = {}
# evaluate the contents of the file
with open(datafile, "r") as fl:
  csvreader = csv.reader(fl, delimiter=",")
  next(csvreader)

  for row in csvreader:
    state = row[4]
    if state in states_count:
      states_count[state] += 1
    else:
      states_count[state] = 1

most_st_customer = max(states_count, key=states_count.get)
print(f'State with most customers: {most_st_customer}'
      f'({states_count[most_st_customer]})')
```

**Question 4** What is the state with the least customers?

```python
# open file and init vars
states_count = {}
# evaluate the contents of the file
with open(datafile, "r") as fl:
  csvreader = csv.reader(fl, delimiter=",")
  header = next(csvreader)

  for row in csvreader:
    if not row:
      continue
    state = row[4].strip()
    state_counts[state] = state_counts.get(state, 0) + 1


min_count = min(state_counts.values())


least_states = [
    state for state, count in state_counts.items() if count == min_count
]

print(f"Minimum customers count: {min_count}")
print(f"States with least customers: {least_states}")
```

**Question #1:** How many unique orders are in the orders.csv file?

**Question #2:** What is the average number of items per order (rounded to two decimal places)?

**Question #3:** What is the highest number of items per order?

**Question #4:** What is the number of orders placed in October 2021?

**Question #5:** Which customer spent the most amount of money in 2021?

**Question #6:** Historically, what is the best month for sales?

```python
# open and evaluate orders.csv

data_orders = "./sample_data/orders.csv"

orders_data= []

with open(data_orders, "r") as fl:
  csvreader = csv.reader(fl, delimiter=",")
  header = next(csvreader)


  for row in csvreader:
    if not row or len(row) < 6:
      continue

    clean_row= [cell.strip() for cell in row]

    if clean_row[0] == "" or clean_row[1] == "":
      continue

    orders_data.append(clean_row)



print(f"Valid row: {len(orders_data)}")

#Q1
unique_orders = set()
for row in orders_data:
  order_id = row[1]
  unique_orders.add(order_id)

print(f"Unique orders: {len(unique_orders)}")

#Q2
total_items = len(orders_data)
totla_u_orders= len(unique_orders)

avg_item = total_items / totla_u_orders
print(f"Average items per order: { round(avg_item, 2)}")


#Q3
items_per_order = {}

for row in orders_data:
  order_id = row[1]
  items_per_order[order_id] = items_per_order.get(order_id, 0) +1


max_items = max(items_per_order.values())
print(f"Highest number of items per order: {max_items}")

#Q4
oct_21 = set()

for row in orders_data:
  order_id = row[1]
  date = row[2]

  if date.startswith("2021-10"):
    oct_21.add(order_id)


print(f"Orders placed in Oct 2021: {len(oct_21)}")


#Q5

customer_spend_21 = {}

for row in orders_data:
  customer_id = row[0]
  date = row[2]

  if date.startswith('2021'):
    try:
      price = float(row[5])
      customer_spend_21[customer_id] = (customer_spend_21.get(customer_id, 0.0) + price)
    except ValueError:
      continue

top_customer = max(customer_spend_21, key=customer_spend_21.get)

print(f"Top Customer in 2021: {top_customer} (Total:"
      f"${round(customer_spend_21[top_customer], 2)})"
)


#Q6
sales_by_mm =  {}

for row in orders_data:
  date = row[2]
  if len(date) >= 7:
    month = date[5:7]
    try:
      price = float(row[5])
      sales_by_mm[month] = sales_by_mm.get(month, 0.0) + price
    except ValueError:
      continue

best_mm = max(sales_by_mm, key=sales_by_mm.get)
print (f"Best Month historically: {best_mm} (Total sales:"
    f" ${round(sales_by_mm[best_mm], 2)})"

  )
```
